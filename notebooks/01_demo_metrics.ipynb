# Retinal Stabilization Demo Metrics
This notebook loads a synthetic video pair and visualises stability metrics.

```python
import pathlib
import matplotlib.pyplot as plt
from retinal_stab.metrics import stability_index
from retinal_stab.ingest import iter_video_frames
from retinal_stab.utils import set_seed

set_seed(42)
base = pathlib.Path('data/synthetic')
before = sorted(base.glob('*.mp4'))[0]
after = pathlib.Path('data/output') / (before.stem + '_stab.mp4')
print('Before:', before)
print('After:', after)

before_frames = list(iter_video_frames(str(before)))
after_frames = list(iter_video_frames(str(after)))
print('Frames loaded', len(before_frames), len(after_frames))
print('Stability index before:', stability_index(before_frames))
print('Stability index after:', stability_index(after_frames))

def flow_hist(frames):
    values = []
    for i in range(len(frames) - 1):
        values.append(stability_index([frames[i], frames[i + 1]]))
    return values

plt.figure(figsize=(6, 4))
plt.hist(flow_hist(before_frames), bins=20, alpha=0.6, label='Before')
plt.hist(flow_hist(after_frames), bins=20, alpha=0.6, label='After')
plt.legend()
plt.title('Optical flow magnitude distribution')
plt.show()

fig, axes = plt.subplots(1, 2, figsize=(8, 4))
axes[0].imshow(before_frames[0][..., ::-1])
axes[0].set_title('Before')
axes[0].axis('off')
axes[1].imshow(after_frames[0][..., ::-1])
axes[1].set_title('After')
axes[1].axis('off')
plt.tight_layout()
plt.show()
```
